```python

import sys
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Dataset filename
AQI_MERGED_CSV = "AQI_merged.csv"

# Set the input directory
if 'google.colab' in sys.modules:
    print("Running in Google Colab")

    from google.colab import drive
    drive.mount('/content/drive')

    INPUT_DIR = '/content/drive/MyDrive/SIADS_696_Milestone_II/Datasets/'

else:
    print("Running in a local environment")

    INPUT_DIR = os.path.abspath(
        os.path.join(os.getcwd(), "../../Datasets/AQI")
    )

print("Input Directory:", INPUT_DIR)

# Build the full path to the dataset
file_path = os.path.join(INPUT_DIR, AQI_MERGED_CSV)

# Load the dataset
if not os.path.isfile(file_path):
    raise FileNotFoundError(
        f"Cannot find {AQI_MERGED_CSV} at {file_path}. "
        "Check that the file is in the specified folder."
    )

df = pd.read_csv(file_path)

print("Dataset loaded successfully!")
print("Dataset shape:", df.shape)

display(df.head())
```

```python

import os

target = "AQI_merged.csv"
matches = []

for root, dirs, files in os.walk('/content/drive/MyDrive'):
    # Skip folders that are unlikely to contain your dataset
    dirs[:] = [d for d in dirs if d not in {
        '.git', '.ipynb_checkpoints', 'node_modules'
    }]

    for filename in files:
        if filename.lower() == target.lower():
            matches.append(os.path.join(root, filename))

if matches:
    print("Found the dataset:")
    for path in matches:
        print(path)
else:
    print("Dataset not found in MyDrive.")
    print("Check whether the CSV is uploaded to Google Drive.")
```

```python
missing_values = df.isnull().sum()

print(missing_values)
```

```python
missing = df.isnull().sum()
missing = missing[missing > 0].sort_values(ascending=False)

if missing.empty:
    print("No missing values detected in the dataset.")
else:
    plt.figure(figsize=(12, 6))
    plt.bar(missing.index.astype(str), missing.values)
    plt.title("Missing Values by Column")
    plt.xlabel("Columns")
    plt.ylabel("Number of Missing Values")
    plt.xticks(rotation=45, ha="right")
    plt.tight_layout()
    plt.show()
```

```python
year_counts = df["Year"].value_counts().sort_index()

plt.figure(figsize=(14, 6))

plt.plot(year_counts.index, year_counts.values)

plt.title("Number of AQI Observations by Year")
plt.xlabel("Year")
plt.ylabel("Number of Observations")

plt.tight_layout()
plt.show()
```

```python
state_counts = df["State"].value_counts()

print(state_counts)
```

```python
counties_by_state = (
    df.groupby("State")["County"]
      .nunique()
      .sort_values()
)

print(counties_by_state)
```

```python
plt.figure(figsize=(12, 10))

counties_by_state.plot(kind="barh")

plt.title("Number of Counties Represented in the AQI Dataset")
plt.xlabel("Number of Counties")
plt.ylabel("State")

plt.tight_layout()
plt.show()
```

```python
county_counts = (
    df.groupby(["State", "County"])
      .size()
      .reset_index(name="Observations")
)

county_counts.head()
```

```python
county_counts_sorted = county_counts.sort_values("Observations")

county_counts_sorted.head(25)
```

```python
state_year_counts = (
    df.groupby(["State", "Year"])
      .size()
      .reset_index(name="Observations")
)

heatmap_data = state_year_counts.pivot(
    index="State",
    columns="Year",
    values="Observations"
)

heatmap_data = heatmap_data.fillna(0)
```

```python
plt.figure(figsize=(18, 14))

sns.heatmap(
    heatmap_data,
    cmap="Blues"
)

plt.title("AQI Data Coverage by State and Year")
plt.xlabel("Year")
plt.ylabel("State")

plt.tight_layout()
plt.show()
```

```python
yearly_aqi = (
    df.groupby("Year")["Median AQI"]
      .mean()
)

plt.figure(figsize=(14, 6))

plt.plot(
    yearly_aqi.index,
    yearly_aqi.values
)

plt.title("Average Median AQI Over Time")
plt.xlabel("Year")
plt.ylabel("Average Median AQI")

plt.tight_layout()
plt.show()
```

```python
yearly_moderate = (
    df.groupby("Year")["Percent Moderate Days"]
      .mean()
)

plt.figure(figsize=(14, 6))

plt.plot(
    yearly_moderate.index,
    yearly_moderate.values
)

plt.title("Average Percentage of Moderate AQI Days Over Time")
plt.xlabel("Year")
plt.ylabel("Percent of Days")

plt.tight_layout()
plt.show()
```

```python
state_aqi = (
    df.groupby("State")["Median AQI"]
      .mean()
      .sort_values()
)

plt.figure(figsize=(12, 10))

state_aqi.plot(kind="barh")

plt.title("Average Median AQI by State")
plt.xlabel("Average Median AQI")
plt.ylabel("State")

plt.tight_layout()
plt.show()
```

```python
state_year_aqi = (
    df.groupby(["State", "Year"])["Median AQI"]
      .mean()
      .reset_index()
)

aqi_heatmap = state_year_aqi.pivot(
    index="State",
    columns="Year",
    values="Median AQI"
)
```

```python
plt.figure(figsize=(18, 14))

sns.heatmap(
    aqi_heatmap,
    cmap="YlOrRd"
)

plt.title("Average Median AQI by State and Year")
plt.xlabel("Year")
plt.ylabel("State")

plt.tight_layout()
plt.show()
```

```python
coverage_summary = (
    df.groupby("State")
      .agg(
          Observations=("County", "size"),
          Counties=("County", "nunique"),
          First_Year=("Year", "min"),
          Last_Year=("Year", "max"),
          Avg_Days_AQI=("Days with AQI", "mean"),
          Avg_Median_AQI=("Median AQI", "mean")
      )
      .sort_values("Observations", ascending=False)
)

coverage_summary
```
